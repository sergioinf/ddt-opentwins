```python
import numpy as np
import plotly.express as px
import pandas as pd
import time
from datetime import timedelta
```

```python
# Load the data into different dataframes
arm_cpu       = pd.read_csv('arm_cpu.csv')
arm_mem       = pd.read_csv('arm_memory.csv')
arm_network   = pd.read_csv('arm_network.csv')

#Set time as index
arm_cpu['Time'] = [timedelta(seconds=i * 15).__str__() for i in range(len(arm_cpu))]
arm_cpu.set_index('Time', inplace=True)
arm_cpu = arm_cpu.apply(lambda x: x/10000000) # Convert to CPU percentage
arm_cpu["Combined"] = arm_cpu[arm_cpu.keys()].sum(axis=1)

#Set time as index
arm_mem['Time'] = [timedelta(seconds=i * 15).__str__() for i in range(len(arm_mem))]
arm_mem.set_index('Time', inplace=True)
arm_mem = arm_mem.apply(lambda x: x/1000) # Convert to MB
arm_mem["Combined"] = arm_mem[arm_mem.keys()].sum(axis=1)


#Set time as index
arm_network['Time'] = [timedelta(seconds=i * 15).__str__() for i in range(len(arm_network))]
arm_network.set_index('Time', inplace=True)
arm_network = arm_network.apply(lambda x: x/1000000) # Convert to MB
arm_network["Combined"] = arm_network[arm_network.keys()].sum(axis=1)
```

### Combining total measurements into one dataset

```python
# Combine one column from each dataframe into a new dataframe
# Select the desired columns from each dataframe
column_cpu = arm_cpu['Combined']
column_mem = arm_mem['Combined']
column_network = arm_network['Combined']

# Combine the selected columns into a new dataframe
combined_df = pd.concat([column_cpu, column_mem, column_network], axis=1)
combined_df.columns = ['CPU', 'Memory', 'Network']
combined_df['Time'] = combined_df.index
# combined_df.set_index('Time', inplace=True)
# Print the combined dataframe
print(combined_df)

combined_df.to_csv('arm_combined.csv', index=False)
```
