```python
import numpy as np
import plotly.express as px
import plotly.io as pio 
import pandas as pd
import time
from datetime import timedelta
from os import listdir
from os.path import isfile, join
import json
from datetime import datetime
import time
```

```python
#Read a folder with json files and create a csv joining them
folder_path = "prueba"  # Replace with the actual folder path

# Get the list of JSON files in the folder
json_files = [file for file in listdir(folder_path) if file.endswith('.json')]

# Create an empty DataFrame to store the combined data
cpu = {
    "time": [],
    "ditto-connectivity": [],
    "ditto-gateway": [],
    "ditto-nginx": [],
    "ditto-policies": [],
    "ditto-things": [],
    "ditto-thingssearch": [],
    "ot-mongodb": [],
    "mosquitto": []
}

memory = {
    "time": [],
    "ditto-connectivity": [],
    "ditto-gateway": [],
    "ditto-nginx": [],
    "ditto-policies": [],
    "ditto-things": [],
    "ditto-thingssearch": [],
    "ot-mongodb": [],
    "mosquitto": []
}
cpu = pd.DataFrame(cpu)
memory = pd.DataFrame(memory)

columns = ["time", "ditto-connectivity", "ditto-gateway", "ditto-nginx", "ditto-policies", "ditto-things", "ditto-thingssearch", "ot-mongodb", "mosquitto"]

tiempo = timedelta(seconds=0)
# Iterate over each JSON file and read it into a DataFrame


for file in json_files:
    file_path = join(folder_path, file)
    with open(file_path, 'r') as f:
        data = json.load(f)
    
    row_cpu = []
    row_memory = []
    
    row_cpu.append(tiempo)
    row_memory.append(tiempo)
    tiempo += timedelta(seconds=30)
    
    
    for pod in data:
        row_cpu.append(pod["containers"][0]["usage"]["cpu"])
        row_memory.append(pod["containers"][0]["usage"]["memory"])
    
    
    row_cpu = pd.DataFrame([row_cpu], columns=columns)  
    row_memory = pd.DataFrame([row_memory], columns=columns)
    
    cpu = pd.concat([cpu,row_cpu], ignore_index=True, axis=0)
    memory = pd.concat([memory,row_memory], ignore_index=True, axis=0)
    # Append the data to the combined DataFrame
    #cpu = cpu._append(row_cpu, ignore_index=True)
    #memory = memory._append(row_memory, ignore_index=True)
```

```python
cpu.to_csv("arm_cpu.csv", index=False)
memory.to_csv("arm_memory.csv", index=False)
```
